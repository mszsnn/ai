## 消息的类型
> 消息支持的两种格式：
1 json格式
2 对象格式

不同格式中，可以携带一些信息，模型需要支持， 同时： 模型提供商也需要支持， 有的能识别， 有的识别不了，详细需要查看文档的官方说明

```python
from langchain.chat_models import init_chat_model
from dotenv import load_dotenv
from langchain_core.messages import SystemMessage, AIMessage, HumanMessage, ToolMessage
from rich import print as p


load_dotenv()


llm = init_chat_model(
    model="openai/gpt-oss-120b",
    model_provider='openrouter',
    temperature= 0,
)

message = [
    {'role': 'system', 'content': '你是一名教师， 整理下不同同学的问题和答案'},
    {'role': 'user', 'content': '1+1=?', 'name': '张三', 'id': 1},
    {'role': 'user', 'content': '2+2=?', 'name': '李四', 'id': 2},
]

response = llm.invoke(message)

p(response.content)


```

```python
message = [
    SystemMessage('你是一名教师， 整理下不同同学的问题和答案'),
    HumanMessage('1+1=?', name = '张三', id = 1),
    HumanMessage('2+2=?', name = '李四', id = 1)
]

response = llm.invoke(message)

p(response.content)
```
